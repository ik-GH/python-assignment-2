```python
#Check whether a number is finite, infinite, or NaN using math.isfinite(), math.isinf(), and math.isnan().

import math
n = float(input("Enter number: "))
if math.isfinite(n):
    print("finite")
elif math.isinf(n):
    print("infinite")
elif math.isnan(n):
    print("NaN")
```

```python
#Find the sum of square roots of all numbers in a list.

import math
numbers = [4, 9, 16]
s = sum(math.sqrt(i) for i in numbers)
print(s)
```

```python
#Calculate compound interest using math.pow().

import math
p = float(input("principal: "))
r = float(input("Rate: "))
t = float(input("Time: "))
a = p * math.pow((1 + r/100), t)
print("Amount =", a)
```

```python
#Find the area and circumference of a circle using math.pi.

import math
r = float(input("Enter radius: "))
area = math.pi * r * r
circumference = 2 * math.pi * r
print("Area =", area)
print("circumference =", circumference)
```

```python
import math

print("1. Square Root")
print("2. Factorial")
print("3. Log")
print("4. Power")
print("5. Trigonometry")
print("6. GCD")
print("7. LCM")

choice = int(input("Enter your choice: "))

if choice == 1:
    n = int(input("Enter number: "))
    print(math.sqrt(n))

elif choice == 2:
    n = int(input("Enter number: "))
    print(math.factorial(n))

elif choice == 3:
    n = int(input("Enter number: "))
    print(math.log10(n))

elif choice == 4:
    a = int(input("Enter base: "))
    b = int(input("Enter power: "))
    print(math.pow(a, b))

elif choice == 5:
    angle = int(input("Enter angle: "))
    r = math.radians(angle)
    print("Sin =", math.sin(r))
    print("Cos =", math.cos(r))
    print("Tan =", math.tan(r))

elif choice == 6:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print(math.gcd(a, b))

elif choice == 7:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print(math.lcm(a, b))

else:
    print("Invalid Choice")
```
