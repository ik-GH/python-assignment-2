```python
#1. Matplotlib

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 50]

plt.plot(x, y, color='blue' , marker='o' , linewidth=2)
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)

plt.show()
```

```python
# plt.plot() - line charts
# plt.bar() - bar charts
# plt.scatter() - scatter plots
# plt.hist() - histograms
# plt.boxplot() - box plots
# plt.pie() - pie charts
```

```python
## simple customization

plt.figure(figsize=(8,5))
plt.bar(x, y, linestyle='--', color='red', label='Sales')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 2. Seaborn

import seaborn as sns
import matplotlib.pyplot as plt

# Sample dataset
flights = sns.load_dataset('flights')

sns.lineplot(data=flights, x='year', y='passengers')
plt.title('Passengers over Time')
plt.show()
```

```python
# sns.lineplot() - line chart
# sns.barplot() - bar chart
# sns.scatterplot() - scatter plot
# sns.histplot() - histogram
# sns.boxplot() - box plot
# sns.heatmap() - heat map
# sns.countplot() - count plots
```

📊 Short Insights



*   The graph shows a strong upward trend in passenger numbers over time
*   Passenger count increases consistently from around **130** in 1949 to nearly 480 in 1960.


*   The highest passenger volume is recorded in **1960**.
*   Growth accelerates after 1955, indicating increasing demand.

*   No major declines are observed throughout the period.
*   Overall, the data suggests **steady and continuous growth in air travel passengers**


```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.boxplot(data=tips, x='day', y='total_bill', palette='Set2')
plt.title('Bill Amount by Day')
plt.show()
```

```python
#3. Plotly

import plotly.express as px

df = px.data.iris()

fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species')
fig.show()
```

```python
import plotly.express as px

df = px.data.gapminder().query("country == 'India'")

fig = px.line(df, x='year', y='lifeExp', title='life expectancy in india')
fig.update_traces(line=dict(color='green', width=3))
fig.show()
```
